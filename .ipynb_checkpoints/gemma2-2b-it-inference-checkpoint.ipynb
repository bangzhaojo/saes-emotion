```python
import pandas as pd
import textwrap
import csv
import numpy as np
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig, GenerationConfig
import sys
from transformers.generation.utils import GreedySearchDecoderOnlyOutput
import textwrap
from tqdm.notebook import tqdm

DEVICE = 'cuda:1'
DEVICE
```

```python
model_name = "google/gemma-2-2b-it"
# download_path = "/scratch/shu.b/models/"
```

```python
# quantization_config = BitsAndBytesConfig(load_in_8bit=True, llm_int8_enable_fp32_cpu_offload=True)
model = AutoModelForCausalLM.from_pretrained(
    model_name, 
    device_map="auto", 
    torch_dtype=torch.bfloat16, 
    # quantization_config=quantization_config, 
    # use_auth_token=token,
    # cache_dir=download_path
    )

tokenizer = AutoTokenizer.from_pretrained(model_name) #, token=token)

tokenizer.pad_token_id = tokenizer.eos_token_id
tokenizer.padding_side = "left"
```

```python
prompt = """Choose the most appropriate emotion from the following list: joy, sadness, anger, disgust, fear, guilt, and shame. Respond with only the emotion, without any explanation.
Experience: {text} 
Answer: """
```

```python
def generate_batch_responses(data, model, tokenizer, device, template, batch_size=8):
    # Ensure the DataFrame has an 'output' column
    if 'output' not in data.columns:
        data['output'] = None

    # Output CSV file path
    csv_filename = "/shared/3/projects/bangzhao/interpretability/gemma2_2b_isear.csv"

    # Load existing responses from CSV if the file exists
    try:
        existing_data = pd.read_csv(csv_filename)
        data.update(existing_data)  # Update DataFrame with previously saved outputs
    except FileNotFoundError:
        pass  # If the file doesn't exist, continue with the current DataFrame

    # Initialize progress bar
    progress_bar = tqdm(total=len(data), desc="Processing Rows", file=sys.stdout)

    # Start processing in batches
    for i in range(0, len(data), batch_size):
        # Skip rows with existing outputs
        batch = data.iloc[i:i + batch_size]
        batch = batch[batch['output'].isnull()]  # Only process rows without output

        if batch.empty:
            progress_bar.update(batch_size)  # Skip updating for processed rows
            continue

        # Prepare prompts for the batch
        batch_prompts = batch[['SIT']].to_dict('records')
        prompts = [
            template.format(text=record['SIT'])
            for record in batch_prompts]
        # for p in prompts:
        #     print(p)
        #     print('-'*10)
        # break
            
        # Tokenize and process the batch
        encoding = tokenizer(prompts, padding=True, return_tensors="pt")
        input_ids = encoding["input_ids"].to(device)
        attention_mask = encoding["attention_mask"].to(device)
        
        with torch.no_grad():
            batch_responses = model.generate(
                input_ids=input_ids,
                attention_mask=attention_mask,
                pad_token_id=tokenizer.pad_token_id,
                max_new_tokens=8,
                temperature=0.0,
            )

        # Decode and store responses
        outputs = []
        for idx, response in enumerate(batch_responses):
            try:
                decoded_output = tokenizer.decode(response, skip_special_tokens=True)
                # Dynamically handle splitting based on expected format
                if "felt:" in decoded_output:
                    split_response = decoded_output.split("felt")[1].strip()

                else:
                    new_tokens = response[input_ids[idx].shape[0]:]  # Exclude prompt tokens
                    decoded_output = tokenizer.decode(new_tokens, skip_special_tokens=True)
                    split_response = "\n".join(textwrap.wrap(decoded_output))
                    
                wrapped_response = "\n".join(textwrap.wrap(split_response))
                outputs.append(wrapped_response)
            except Exception as e:
                outputs.append("Error decoding response")  # Handle decoding errors gracefully

        # Update 'output' column
        data.loc[batch.index, 'output'] = outputs

        # Save progress after each batch
        data.to_csv(csv_filename, index=False)

        # Update progress bar
        progress_bar.update(len(batch))

    # Close the progress bar
    progress_bar.close()

    # Save the final DataFrame to the CSV file
    data.to_csv(csv_filename, index=False)
    return data
```

```python
isear = pd.read_csv("/shared/3/projects/bangzhao/interpretability/isear_clean.csv")
isear = isear[~isear.apply(lambda row: row['Field1'].lower() in row['SIT'].lower(), axis=1)]
isear = isear[['ID', 'SIT', 'Field1']]
```

```python
data_output = generate_batch_responses(
    data=isear,
    model=model,
    tokenizer=tokenizer,
    device=DEVICE,
    template = prompt,
    batch_size=16
)
```
